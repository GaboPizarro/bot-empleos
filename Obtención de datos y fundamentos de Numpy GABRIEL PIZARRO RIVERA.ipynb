```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("06. Apoyo desafío - Datos climáticos.csv")
#print(df.columns)

datos_numericos = df[["temperatura_celsius", "humedad_relativa"]].to_numpy()
print(datos_numericos)
```

```python
# a. Selección las primeras 12 filas
primeras_12 = datos_numericos[:12]
print("a. Primera hora de mediciones (primeras 12 filas):")
print(primeras_12)
```

```python
# b. Encuentra todas las temperaturas superiores a 25.0 °C.
temperaturas = datos_numericos[:, 0]
temperaturas_altas = temperaturas[temperaturas > 25.0]
print("b. Temperaturas superiores a 25.0 °C:")
print(temperaturas_altas)
```

```python
# c. Convierte temperatura de Celsius a Fahrenheit.
# fórmula F = C * (9/5) + 32.
temperatura_fahrenheit = (temperaturas * 9 / 5) + 32
print("c. Temperaturas en Fahrenheit (primeras 5):")
print(temperatura_fahrenheit[:5])
```

```python
# d. Crea una matriz de 2x2 con la primera y la última medición de temperatura y humedad del día.
matriz_2x2 = np.array([datos_numericos[0], datos_numericos[-1]])
print("d. Matriz 2x2 (primera y última medición del día):")
print(matriz_2x2)
```

```python
# e. Guardar temperatura_fahrenheit en un archivo CSV
df_fahr = pd.DataFrame({'temperatura_fahrenheit': temperatura_fahrenheit})
df_fahr.to_csv('temperaturas_fahrenheit.csv', index=False)
print(f"e. Archivo 'temperaturas_fahrenheit.csv' creado con {len(df_fahr)} registros")
```
